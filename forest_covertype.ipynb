### handel outliers

```python
num_cols = [
    "Elevation", "Aspect", "Slope",
    "Horizontal_Distance_To_Hydrology", "Vertical_Distance_To_Hydrology",
    "Horizontal_Distance_To_Roadways",
    "Hillshade_9am", "Hillshade_Noon", "Hillshade_3pm",
    "Horizontal_Distance_To_Fire_Points",
]

df_clean = df.copy()
for col in num_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df_clean = df_clean[(df_clean[col] >= lower) & (df_clean[col] <= upper)]

print("before", df.shape)
print("After", df_clean.shape)
```

```python
df_clean.Cover_Type.value_counts()
```

لما جربت احذف الاوتليرس لقيت انه الداتا اختلفت اوى وبالاخص فى التارجت وده مينفعش واصلا دى قيم ارتفاعات وحاجات طبيعيه يعنى  
هتاثر عالداتا لو اتحذفت عشان كده مش هحذف حاجه وهستعمل الكاب عشان المودل الحساسه للاوتليرس  

```python
df_cap = df.copy()
for col in num_cols:
    lower = df_cap[col].quantile(0.01)
    upper = df_cap[col].quantile(0.99)
    df_cap[col] = df_cap[col].clip(lower, upper)
```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

scaler = StandardScaler()
x_train_s = scaler.fit_transform(x_train)
x_test_s = scaler.transform(x_test)

x_small, _, y_small, _ = train_test_split(
    x_train_s, y_train, train_size=0.1, stratify=y_train, random_state=42)

params = {'n_neighbors': [3, 5, 9, 15, 21],
          'weights': ['uniform', 'distance']}

grid_search = GridSearchCV(
    KNeighborsClassifier(n_jobs=-1),
    params, cv=3, scoring='accuracy', n_jobs=-1)
grid_search.fit(x_small, y_small)

print("Best params:", grid_search.best_params_)
print("CV accuracy:", grid_search.best_score_)

knn_best = KNeighborsClassifier(**grid_search.best_params_, n_jobs=-1)
knn_best.fit(x_train_s, y_train)

n = 20000
y_pred = knn_best.predict(x_test_s[:n])
print("Accuracy:", accuracy_score(y_test[:n], y_pred))
```

```python
x_tr, x_val, y_tr, y_val = train_test_split(
    x_small, y_small, test_size=0.2, random_state=42)

acc = []
ks = range(1, 30, 2)
for k in ks:
    knn = KNeighborsClassifier(n_neighbors=k, n_jobs=-1)
    knn.fit(x_tr, y_tr)
    acc.append(accuracy_score(y_val, knn.predict(x_val)))

plt.plot(ks, acc, marker='o')
plt.xlabel('k')
plt.ylabel('Accuracy')
plt.show()
```

### DT


```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import classification_report, accuracy_score, roc_auc_score
from sklearn import tree

dt = DecisionTreeClassifier(random_state=42)
dt.fit(x_train, y_train)

y_pred_dt = dt.predict(x_test)
y_proba_dt = dt.predict_proba(x_test)

print("Decision Tree (baseline)")
print(classification_report(y_test, y_pred_dt))
print("Train Accuracy:", dt.score(x_train, y_train))
print("Test Accuracy:", accuracy_score(y_test, y_pred_dt))
print("ROC-AUC (ovr):", roc_auc_score(y_test, y_proba_dt, multi_class='ovr'))

param_grid = {
    'max_depth': [10, 20, 30, None],
    'min_samples_leaf': [1, 5, 10],
    'criterion': ['gini', 'entropy']
}

grid = GridSearchCV(
    DecisionTreeClassifier(random_state=42),
    param_grid, scoring='accuracy', cv=3, n_jobs=-1)
grid.fit(x_train, y_train)

best_dt = grid.best_estimator_
print("Best params:", grid.best_params_)
print("Best CV accuracy:", grid.best_score_)

y_pred_best = best_dt.predict(x_test)
print("Test Accuracy (best):", accuracy_score(y_test, y_pred_best))
print(classification_report(y_test, y_pred_best))
print("ROC-AUC (ovr):", roc_auc_score(y_test, y_proba_dt, multi_class='ovr'))
imp = pd.Series(best_dt.feature_importances_, index=x_train.columns)
imp.sort_values(ascending=False).head(10).plot(kind='barh')
plt.gca().invert_yaxis()
plt.title("Top 10 Feature Importances")
plt.show()

small_tree = DecisionTreeClassifier(max_depth=3, random_state=42).fit(x_train, y_train)
plt.figure(figsize=(20, 10))
tree.plot_tree(small_tree, filled=True, feature_names=x_train.columns,
               class_names=[str(c) for c in small_tree.classes_], fontsize=9)
plt.show()
```

### RF

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV, train_test_split
from sklearn.metrics import classification_report, accuracy_score, roc_auc_score

rf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf.fit(x_train, y_train)

print('Train accuracy:', rf.score(x_train, y_train))
print('Test accuracy:', rf.score(x_test, y_test))
print(classification_report(y_test, rf.predict(x_test)))

x_small, _, y_small, _ = train_test_split(
    x_train, y_train, train_size=0.15, stratify=y_train, random_state=42)

param_dist = {
    'n_estimators': [100, 200],
    'max_depth': [20, 30, None],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 0.5]
}

search = RandomizedSearchCV(
    RandomForestClassifier(random_state=42, n_jobs=-1),
    param_dist, n_iter=8, scoring='accuracy', cv=3,
    random_state=42, n_jobs=1)
search.fit(x_small, y_small)

print("Best params:", search.best_params_)
print("Best CV accuracy:", search.best_score_)

best_rf = RandomForestClassifier(**search.best_params_, random_state=42, n_jobs=-1)
best_rf.fit(x_train, y_train)

y_pred_train = best_rf.predict(x_train)
y_pred = best_rf.predict(x_test)

print('Train accuracy:', accuracy_score(y_train, y_pred_train))
print('Test accuracy:', accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
print("ROC-AUC (ovr):", roc_auc_score(y_test, best_rf.predict_proba(x_test), multi_class='ovr'))
```

### AdaBoost

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score

ada = AdaBoostClassifier(n_estimators=50, random_state=42)
ada.fit(x_train, y_train)

y_pred_ada = ada.predict(x_test)
print(f"AdaBoost (default) accuracy: {accuracy_score(y_test, y_pred_ada):.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred_ada))
ada_deep = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=10, random_state=42),
    n_estimators=50,
    learning_rate=0.5,
    random_state=42)
ada_deep.fit(x_train, y_train)

y_pred_deep = ada_deep.predict(x_test)
print(f"AdaBoost (deep base) accuracy: {accuracy_score(y_test, y_pred_deep):.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred_deep))

y_proba = ada_deep.predict_proba(x_test)
print("ROC-AUC (ovr):", roc_auc_score(y_test, y_proba, multi_class='ovr'))
```

### XGB

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score

y_train_x = y_train - 1
y_test_x = y_test - 1

xgb_model = XGBClassifier(
    n_estimators=300,
    learning_rate=0.2,
    max_depth=12,
    subsample=0.8,
    colsample_bytree=0.8,
    tree_method="hist",
    eval_metric="mlogloss",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(x_train, y_train_x)

y_pred_xgb = xgb_model.predict(x_test)
print(f"XGBoost accuracy: {accuracy_score(y_test_x, y_pred_xgb):.4f}")
print("Classification Report:")
print(classification_report(y_test_x, y_pred_xgb))

y_proba_xgb = xgb_model.predict_proba(x_test)
print("ROC-AUC (ovr):", roc_auc_score(y_test_x, y_proba_xgb, multi_class='ovr'))
```

```python
## We selected XGBoost for hyperparameter tuning because it achieved the highest
## initial test accuracy (97.05%) among the evaluated models, with a macro F1-score
## of 0.94 and a macro-averaged one-vs-rest ROC-AUC of 0.999. Although Random Forest
## performed similarly, XGBoost offers more tunable hyperparameters (e.g., learning
## rate, tree depth, subsampling), so we used GridSearchCV to find the best
## combination of hyperparameters to improve its accuracy and macro F1-score.
```

```python
from xgboost import XGBClassifier
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.metrics import (accuracy_score, classification_report,
                             roc_auc_score, ConfusionMatrixDisplay)

y_train_x = y_train - 1
y_test_x = y_test - 1

x_small, _, y_small, _ = train_test_split(
    x_train, y_train_x, train_size=0.15, stratify=y_train_x, random_state=42)

param_grid = {
    'max_depth': [8, 12],
    'learning_rate': [0.1, 0.2],
    'n_estimators': [200, 300],
    'subsample': [0.8]
}

grid_xgb = GridSearchCV(
    XGBClassifier(tree_method='hist', eval_metric='mlogloss',
                  colsample_bytree=0.8, random_state=42, n_jobs=-1),
    param_grid, scoring='accuracy', cv=3, n_jobs=1, verbose=1)
grid_xgb.fit(x_small, y_small)

print("Best Parameters:", grid_xgb.best_params_)
print("Best CV Accuracy:", grid_xgb.best_score_)

best_xgb = XGBClassifier(**grid_xgb.best_params_, tree_method='hist',
                         eval_metric='mlogloss', colsample_bytree=0.8,
                         random_state=42, n_jobs=-1)
best_xgb.fit(x_train, y_train_x)

y_pred_best = best_xgb.predict(x_test)
y_proba_best = best_xgb.predict_proba(x_test)

print("Test Accuracy:", accuracy_score(y_test_x, y_pred_best))
print(classification_report(y_test_x, y_pred_best))
print("ROC-AUC (ovr):", roc_auc_score(y_test_x, y_proba_best, multi_class='ovr'))

fig, ax = plt.subplots(figsize=(8, 8))
ConfusionMatrixDisplay.from_predictions(
    y_test_x, y_pred_best,
    display_labels=[f'Class {i}' for i in range(1, 8)],
    normalize='true', ax=ax)
plt.show()
```
